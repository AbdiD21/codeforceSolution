### Problem 25A: IQ test

The link to the problem is https://codeforces.com/problemset/problem/25/A


#### **Solution**

```python
n = int(input())  # Number of elements
numbers = list(map(int, input().split()))  # List of numbers

# Create a list of the parity (0 if even, 1 if odd) for each number
parity = []
for num in numbers:
    parity.append(num % 2)
# Alternatively, you can use a list comprehension to create the parity list
# parity = [num % 2 for num in numbers]

# Check which parity is the minority and return its index + 1 (1-based indexing)
if parity.count(0) == 1:
    print(parity.index(0) + 1)
else:
    print(parity.index(1) + 1)
```

### Python **methods** used in this problem

##### **Python string split() method**

The `split() method` breaks down/splits a string into a list of substrings `using` a chosen spearator(also called a delimiter).

It is very useful when you're working with text and want to break it into parts.

***Basic split() syntax***

        string.split(separator, maxsplit)

The split() method takes a maximum of 2 parameters:

- separator (optional) - Specifies the separator/delimiter `used to` split the string. If not provided, whitespace is used as the default separator/delimiter.

- maxsplit (optional) - Determines the maximum number of splits. If not provided, the default value is -1, which means there is no limit on the number of splits.

##### Examples

1. Split by spaces (default behavior)


```python
sentence = "progamming is easy"
words = sentence.split()
print(words)
```

2. Split using a comma as a separator

```python
data = "apple,banana,mango"
fruits = data.split(",")
print(fruits)
```

3. Split with a max number of splits

```python
numbers = "one two three four"
parts = numbers.split(" ", 2)
print(parts) # Only the first 2 spaces are used to split.
```

Only the first 2 spaces are used to split.

4. Split using a custom delimiter

```python
info = "name|age|country"
pieces = info.split("|")
print(pieces)
```

5. Split a sentence into words and loop over them

```python
sentence = "Learn Python step by step"
for word in sentence.split():
    print(word)
```

##### **Python List count()**

- The count() method returns the number of times the specified element appears in the list.

- The count() method in Python is used to count the ***number of occurrences*** of a specified value in a list. It `returns` an integer representing `how many times` the value appears in the list.


**Syntax**

        list.count(value) 
              or 
        list.count(element)
- value: The value to count in the list(element - the element to be counted)

##### Examples

1.  how to use the count() method

```python
# Creating a list
fruits = ['mango', 'banana', 'apple', 'banana']

# Counting occurrences of 'banana'
banana_count = fruits.count('banana')
print(banana_count)  # Output: 2
```

2. Counting Other Values

You can count any type of value, including numbers and strings:

```python
# Counting occurrences of a number
numbers = [1, 2, 3, 2, 4, 2]
two_count = numbers.count(2)
print(two_count)
```

3. Counting Non-Existent Values

If the specified value is not found in the list, count() will return 0:

```python
# Counting a value not in the list
orange_count = fruits.count('orange')
print(orange_count)
```

4. Example with Mixed Data Types

You can also count occurrences in a list with mixed data types:

```python
mixed_list = [1, 'apple', 2, 'banana', 1, 'apple']
apple_count = mixed_list.count('apple')
print(apple_count)
```

##### **Python List index()**

- The **index()** method returns the index of the specified element in the list.

- The **index() method** in Python is used `to find` the **first occurrence** of a specified value in a list and returns its position (index). 

- If the value is not found, it raises a `ValueError`.

**Syntax**

        list.index(element, start, end)

The list index() method can take a maximum of three arguments:

- element - the element to be searched
- start (optional) - start searching from this index
- end (optional) - search the element up to this index

##### Examples

1. simple example demonstrating how to use the `index()` method:

```python
# Creating a list
fruits = ['apple', 'banana', 'cherry', 'date', 'banana']

# Finding the index of 'banana'
index_of_banana = fruits.index('banana')
print(index_of_banana) 
```

2. Using start and end Parameters

You can also specify the start and end parameters to limit the search:

```python
# Finding the index of 'banana' starting from index 2
index_of_banana_from_2 = fruits.index('banana', 2)
print(index_of_banana_from_2)
```

3. Handling ValueError

If the specified value is not found in the list, a `ValueError`  is raised. It's good practice to handle this exception:

```python
try:
    index_of_orange = fruits.index('orange')
except ValueError:
    print("Value not found in the list.")
```

***Example with a Custom List***

Let's see a more complex example with a list of numbers:

```python
numbers = [10, 20, 30, 40, 50, 60]

# Finding the index of 30
index_of_30 = numbers.index(30)
print(index_of_30)  # Output: 2

# Trying to find a number that doesn't exist
try:
    index_of_100 = numbers.index(100)
except ValueError:
    print("100 is not in the list.")
```
