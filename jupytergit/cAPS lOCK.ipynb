### **A.cAPS lOCK**

Problem link: https://codeforces.com/problemset/problem/131/A

#### What this code will handle:
- Full uppercase words → lowercased.
- First letter lowercase, rest uppercase → proper capitalization.
- Single lowercase letter → uppercased.
- All other cases → unchanged.


#### **Code**

````python
# Read the input word from the user
word = input()

# Case 1: If all characters are uppercase (e.g., "LOCK")
# That means the whole word is in caps by mistake, so we convert it to lowercase
if word.isupper():
    print(word.lower())

# Case 2: If all characters except the first are uppercase (e.g., "cAPS")
# We need to capitalize the first letter and lowercase the rest
elif len(word) > 1 and word[1:].isupper():
    # word[0].upper() capitalizes the first letter
    # word[1:].lower() lowercases the rest of the string
    print(word[0].upper() + word[1:].lower())

# Case 3: Special case — if the word is just a single lowercase letter like "a"
# We convert it to uppercase
elif len(word) == 1 and word.islower():
    print(word.upper())

# Case 4: If none of the above conditions are true
# Then the word was typed correctly — we print it as is
else:
    print(word)
````

#### **Time and Space Complexity**

### Time Complexity

Let **n = length of the string `word`**.

| Operation            | Complexity |
| -------------------- | ---------- |
| `word.isupper()`     | O(n)       |
| `word[1:].isupper()` | O(n)       |
| `word.lower()`       | O(n)       |
| `word[1:].lower()`   | O(n)       |
| `word[0].upper()`    | O(1)       |

**Total time complexity: O(n)**
Why? Because the string is checked and possibly copied/modified a few times — all linear operations.


### Space Complexity

| What’s stored              | Complexity |
| -------------------------- | ---------- |
| Input string `word`        | O(n)       |
| Possibly new string copies | O(n)       |

**Total space complexity: O(n)**
Even though the space used is small, **string operations like `.lower()` and `.upper()` create new strings** (since strings are immutable in Python).


### Generally :

* **Time Complexity:** `O(n)`
* **Space Complexity:** `O(n)`
