### 1A.Theatre Square

problem link: https://codeforces.com/problemset/problem/1/A

**Problem Details**

We have a rectangle of size n × m.

We want to cover it with square tiles of size a × a.

Tiles can hang over the edge, but cannot be broken.

* Find minimum number of tiles required to cover the whole rectangle.

***Input Format:***

Single line with three integers:

```python
n m a
```
Where:
   * n = height of the square

   * m = width of the square

   * a = side length of a tile

**Output:**

Print the minimum number of tiles required.


Logic:

We need:

   * `ceil(n / a)` tiles along height

   * `ceil(m / a)` tiles along width

Total tiles = `ceil(n / a) * ceil(m / a)`

In Python, to avoid float division:
```python
      tiles = ((n + a - 1) // a) * ((m + a - 1) // a)
```

##### Code

````python
n, m, a = map(int, input().split())
tiles = ((n + a - 1) // a) * ((m + a - 1) // a)
print(tiles)
````
