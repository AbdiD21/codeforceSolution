### B. The New Year: Meeting Friends

The algorithm to solve the "Meeting Friends" problem is:

1. Input: Read three distinct integers x1, x2, and x3 representing the coordinates of the three friends.

2. Sort: Sort the three integers in ascending order. Let the sorted values be a, b, and c, where a is the smallest, b is the median, and c is the largest.

3. Calculate Distance: Calculate the difference between the largest and smallest values: distance = c - a.

4. Output: Print the calculated distance.

The optimal meeting point is the median of the three locations. The minimum total distance is the difference between the largest and smallest of the three locations. This is because the total distance is minimized when the meeting point is the median.


**_Implementation_**


```python
def solve():
    x1, x2, x3 = map(int, input().split())
    # sorted() is a built-in method in Python.
    arr = sorted([x1, x2, x3])
    print(arr[2] - arr[0])


solve()
```

The problem requires finding the minimum total distance three friends have to travel to meet at one point on a line. The optimal meeting point is always the median of the three locations. The minimum total distance is the difference between the largest and smallest of the three locations.

Given three points x1, x2, and x3, the math is as follows:

1. Find the maximum and minimum of the three points.

2. Calculate the difference between the maximum and minimum values.

For example, if x1 = 7, x2 = 1, and x3 = 4:

1. The sorted points are 1, 4, and 7.

2. The minimum is 1, and the maximum is 7.

3. The minimum total distance is 7 - 1 = 6.


Time and Space Complexity

Time Complexity: O(1).
The sorting operation takes constant time because we are always sorting an array of size 3.

Space Complexity: O(1).
The algorithm uses a fixed amount of extra space to store the sorted array and a few variables, regardless of the input size.
