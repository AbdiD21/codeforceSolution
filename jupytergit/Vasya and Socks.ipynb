### **460. Vasya and Socks**

The  link to the problem: https://codeforces.com/problemset/problem/460/A 


### **Approach**


* Vasya is a student.

* Every morning, he wears 1 pair of socks to go to school.

* When he comes back, he throws the socks away (so he can't reuse them).

* But! His mom helps him by buying 1 new pair of socks every m-th day (like on day m, 2m, 3m, ...).

**Goal:**

* Find out how many days Vasya can go to school before he runs out of socks.

**Input:**

You are given:
 - n: number of socks he starts with.
 - m: his mom gives him 1 new sock every m days.
```python
                   Input: 9 3
```
This means:

  - Vasya has 9 pairs of socks.

  - His mom gives him 1 extra pair every 3 days.

**Output:**

* Print the total number of days Vasya can go to school before he completely runs out of socks.

For example:
```python
        Input: 9 3
        Output: 13
```

How it works (Example with n=9, m=3):

| Day | Socks Left | New Sock? | Total Days |
| --- | ---------- | --------- | ---------- |
| 1   | 9 → 8      | ❌         | 1          |
| 2   | 8 → 7      | ❌         | 2          |
| 3   | 7 → 6+1=7  | ✅ +1      | 3          |
| 4   | 7 → 6      | ❌         | 4          |
| 5   | 6 → 5      | ❌         | 5          |
| 6   | 5 → 4+1=5  | ✅ +1      | 6          |
| 7   | 5 → 4      | ❌         | 7          |
| 8   | 4 → 3      | ❌         | 8          |
| 9   | 3 → 2+1=3  | ✅ +1      | 9          |
| 10  | 3 → 2      | ❌         | 10         |
| 11  | 2 → 1      | ❌         | 11         |
| 12  | 1 → 0+1=1  | ✅ +1      | 12         |
| 13  | 1 → 0      | ❌         | 13         |
* On day 14, he has 0 socks. So he stops at day 13.

Conclusion:
* Vasya uses 1 sock per day.

* His mom helps every m-th day with 1 new sock.

* You simulate day by day and count until socks are gone.
